```python
cols = ['Restaurant Name', 'City', 'Locality', 'Cuisines']

df[cols] = df[cols].apply(lambda x: x.str.title())
```

```python
df[['Restaurant Name','City','Locality','Cuisines']].head()
```

```python
cols = ['Restaurant Name', 'City', 'Locality', 'Cuisines']

df[cols] = df[cols].apply(lambda x: x.str.strip().str.replace(r'\s+', ' ', regex=True))
```

```python
print(df[['Restaurant Name', 'City', 'Locality', 'Cuisines']].head())
```

```python
binary_cols = []

for col in df.columns:
    if set(df[col].dropna().unique()) == {'Yes', 'No'}:
        binary_cols.append(col)

print(binary_cols)
```

```python
binary_cols = ['Has Table booking', 'Has Online delivery', 'Is delivering now']

for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})
```

```python
print(df[binary_cols].head())
```

```python
df = pd.get_dummies(df, columns=['Currency', 'Rating text'])
```

```python
print(df.head())
```

```python
cols = ['Average Cost for two', 'Votes', 'Aggregate rating']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\n{col}")
    print("Number of outliers:", len(outliers))
    print(outliers[[col]].head())
```

```python
cols = ['Average Cost for two', 'Votes', 'Aggregate rating']

df_clean = df.copy()

for col in cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean = df_clean[(df_clean[col] >= lower) & (df_clean[col] <= upper)]
```

```python
print("Original dataset size:", df.shape)
print("Dataset size after removing outliers:", df_clean.shape)
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

ss = StandardScaler()
mms = MinMaxScaler()

df['Votes'] = ss.fit_transform(df[['Votes']])

df['Average Cost for two'] = mms.fit_transform(df[['Average Cost for two']])
```

```python
print(df[['Votes', 'Average Cost for two']].head())
```

```python
df['Cost Per Person'] = df['Average Cost for two'] / 2
```

```python
df['High Rated Restaurant'] = (df['Aggregate rating'] >= 4.0).astype(int)
```

```python
df['Popular Restaurant'] = (df['Votes'] > 100).astype(int)
```
